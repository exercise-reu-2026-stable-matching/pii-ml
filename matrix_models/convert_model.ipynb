```python
from collections import OrderedDict
import torch
```

```python
# 1. Load the state dictionary from your file
model_path = "saved_transformer_models/matrixStateData_2000000_20_iter0_hs64/matrixStateData_2000000_20_iter0_IDddp2064128_ep0070"
state_dict = torch.load(model_path + ".pt")

# 2. Create a new OrderedDict to store the corrected keys
new_state_dict = OrderedDict()
for k, v in state_dict.items():
    # Remove the "module." prefix
    name = k.replace("module.", "") if k.startswith("module.") else k
    new_state_dict[name] = v

torch.save(new_state_dict, model_path + "_converted.pt")
```
