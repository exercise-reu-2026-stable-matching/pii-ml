```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def plot_data(
        x, ys: np.ndarray, labels: list[str] | None = None, title: str = "", ylabel: str = ""
    ) -> tuple:
    fig, ax = plt.subplots()

    if len(ys.shape) > 1:
        assert isinstance(labels, list)
        for y in ys:
            ax.plot(x, y)
    else:
        line = ax.plot(x, ys)
        ax.legend(handles=line)

    ax.set(xlabel="Epoch", ylabel=ylabel, title=title)
    ax.grid()

    if labels != None:
        ax.legend(labels)

    return fig, ax
```

```python
data_names = [
    "matrixStateData_2000000_20_iter0_ID105217",
]

data_name = data_names[0]

dfs = [pd.read_csv(f"transformer_matrix_plot_data/{data_name}.csv") for data_name in data_names]
df = pd.concat(dfs)
df
```

```python
x = df["epochs"]

losses = np.vstack((df["train_loss"], df["test_loss"]))
loss_fig, loss_ax = plot_data(x, losses, ["Training", "Testing"], f"Loss vs. Epoch ({data_name})", "Loss")
# plt.savefig(f"transformer_matrix_plots/{data_name}_iter{iter_index}_ID{JOB_ID}_loss")

accuracies = np.vstack((df["train_accuracy"], df["test_accuracy"])) * 100
acc_fig, acc_ax = plot_data(x, accuracies, ["Training", "Testing"], f"Accuracy vs. Epoch ({data_name})", "Accuracy (%)")
# plt.savefig(f"transformer_matrix_plots/{data_name}_iter{iter_index}_ID{JOB_ID}_acc")

lr_fig, lr_ax = plot_data(x, df["learning_rate"].to_numpy(), title=f"Learning Rate vs. Epoch ({data_name})", ylabel="Learning Rate")
# plt.savefig(f"transformer_matrix_plots/{data_name}_iter{iter_index}_ID{JOB_ID}_lr")
```
