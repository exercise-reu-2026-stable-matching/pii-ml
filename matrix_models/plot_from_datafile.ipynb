```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def plot_data(
        x, ys: np.ndarray, labels: list[str] | None = None, title: str = "", ylabel: str = ""
    ) -> tuple:
    fig, ax = plt.subplots()

    if len(ys.shape) > 1:
        assert isinstance(labels, list)
        for y in ys:
            ax.plot(x, y)
    else:
        line = ax.plot(x, ys)
        ax.legend(handles=line)

    ax.set(xlabel="Epoch", ylabel=ylabel, title=title)
    ax.grid()

    if labels != None:
        ax.legend(labels)

    return fig, ax
```

```python
data_names = [
    "matrixStateData_2000000_20_iter0_IDddp2064128",
    "matrixStateData_2000000_20_iter0_ID105349",
]

data_name = data_names[0]

dfs = [pd.read_csv(f"transformer_matrix_plot_data/{data_name}.csv") for data_name in data_names]
df = pd.concat(dfs)
```

```python
x = df["epochs"]

losses = np.vstack((df["train_loss"], df["test_loss"]))
loss_fig, loss_ax = plot_data(x, losses, ["Training", "Testing"], f"Loss vs. Epoch ({data_name})", "Loss")
# plt.savefig(f"transformer_matrix_plots/{data_name}_iter{iter_index}_ID{JOB_ID}_loss")

accuracies = np.vstack((df["train_accuracy"], df["test_accuracy"])) * 100
acc_fig, acc_ax = plot_data(x, accuracies, ["Training", "Testing"], f"Accuracy vs. Epoch ({data_name})", "Accuracy (%)")
# plt.savefig(f"transformer_matrix_plots/{data_name}_iter{iter_index}_ID{JOB_ID}_acc")

lr_fig, lr_ax = plot_data(x, df["learning_rate"].to_numpy(), title=f"Learning Rate vs. Epoch ({data_name})", ylabel="Learning Rate")
# plt.savefig(f"transformer_matrix_plots/{data_name}_iter{iter_index}_ID{JOB_ID}_lr")
```

```python
data_names = [
    "matrixStateData_2000000_10_iter0_ID105100_n10_hs64_bs64_pt1",
    "matrixStateData_2000000_10_iter0_ID105497_n10_hs64_bs64_masked",
]

dfs = [pd.read_csv(f"transformer_matrix_plot_data/{data_name}.csv") for data_name in data_names]

van_df, mask_df = dfs
```

```python
display(van_df, mask_df)
```

```python
fig, ax = plt.subplots()

num_epochs = max(len(van_df["epochs"]), len(mask_df["epochs"]))
x = np.arange(num_epochs)

van_train_acc = van_df["train_accuracy"] * 100
van_test_acc = van_df["test_accuracy"] * 100

mask_train_acc = mask_df["train_accuracy"] * 100
mask_test_acc = mask_df["test_accuracy"] * 100

ax.plot(x, van_train_acc, color="#E65B39", linestyle="dashed", alpha=0.5)
ax.plot(x, van_test_acc, color="#E65B39")
ax.plot(x, mask_train_acc, color="#5EBD37", linestyle="dashed", alpha=0.5)
ax.plot(x, mask_test_acc, color="#5EBD37")

ax.set(xlabel="Epoch", ylabel="Accuracy (%)", title="Masked/Unmasked Accuracy vs. Epoch")
ax.grid()

ax.legend(["Unmasked Train Accuracy", "Unmasked Test Accuracy", "Masked Train Accuracy", "Masked Test Accuracy"])

plt.savefig("transformer_matrix_plots/masked_vs_unmasked.png", dpi=400)
```
